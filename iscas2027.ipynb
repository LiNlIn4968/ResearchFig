```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --- 1. 读取数据并平移红、黄色曲线频率 ---
df = pd.read_excel('cmrr.xlsx')
df['Freq_Unadjusted_shifted'] = df['Freq_Unadjusted'] / 10.0
df['Freq_Adjusted_shifted'] = df['Freq_Adjusted'] / 10.0

x_adm = df['dB20(VF("/VOUTP")) (Vctrl=1.48) X'].dropna()
y_adm = df['dB20(VF("/VOUTP")) (Vctrl=1.48) Y'].dropna()
x_unadj = df['Freq_Unadjusted_shifted'].dropna()
y_unadj = df['Gain_Unadjusted_vctrl0.8'].dropna()
x_adj = df['Freq_Adjusted_shifted'].dropna()
y_adj = df['Gain_Adjusted_vctrl1.2'].dropna()

# 计算 50Hz 处的内插值及 CMRR
val_adm = np.interp(np.log10(50), np.log10(x_adm), y_adm)
val_unadj = np.interp(np.log10(50), np.log10(x_unadj), y_unadj)
val_adj = np.interp(np.log10(50), np.log10(x_adj), y_adj)

cmrr_unadj = val_adm - val_unadj
cmrr_adj = val_adm - val_adj
improvement = cmrr_adj - cmrr_unadj

# --- 2. IEEE 顶刊样式设置 ---
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman', 'DejaVu Serif']
plt.rcParams['mathtext.fontset'] = 'stix'

plt.rcParams['axes.linewidth'] = 1.0
plt.rcParams['xtick.direction'] = 'in'
plt.rcParams['ytick.direction'] = 'in'
plt.rcParams['xtick.top'] = True
plt.rcParams['ytick.right'] = True
plt.rcParams['xtick.minor.visible'] = True
plt.rcParams['ytick.minor.visible'] = True

# 字号设置
plt.rcParams['font.size'] = 11
plt.rcParams['axes.labelsize'] = 12
plt.rcParams['legend.fontsize'] = 10
plt.rcParams['xtick.labelsize'] = 11
plt.rcParams['ytick.labelsize'] = 11

fig, ax = plt.subplots(figsize=(6.5, 4.2), dpi=300)

# --- 3. 绘制曲线 ---
curves = [
    (x_adm, y_adm, 'Adm', '#0000FF'),
    (x_unadj, y_unadj, 'Acm (W/O NC trimming)', '#FF0000'),
    (x_adj, y_adj, 'Acm (W/ NC trimming)', '#D8C61C')
]

for x_col, y_col, label, color in curves:
    ax.semilogx(x_col, y_col, color=color, label=label, linewidth=1.8)

# --- 4. 50Hz 处的垂线与双箭头标注 ---
ax.axvline(x=50, color='#000000', linestyle='--', linewidth=1.0, alpha=0.9)

# 标记 50Hz 处的数据点
ax.plot(50, val_adm, 'o', color='#0000FF', markersize=5, zorder=7)
ax.plot(50, val_unadj, 'o', color='#FF0000', markersize=5, zorder=7)
ax.plot(50, val_adj, 'o', color='#D8C61C', markersize=5, zorder=7)

# 第一个箭头：未调谐 CMRR (Adm 到 Acm_unadjusted)
ax.annotate('', xy=(30, val_unadj), xytext=(30, val_adm),
            arrowprops=dict(arrowstyle='<->', color='#FF0000', lw=1.5, mutation_scale=10))
ax.text(25, (val_adm + val_unadj) / 2, rf'CMRR: {cmrr_unadj:.1f} dB@50Hz',
        color='#FF0000', fontsize=9, fontweight='bold', ha='right', va='center')

# 第二个箭头：已调谐 CMRR (Adm 到 Acm_adjusted)
ax.annotate('', xy=(100, val_adj), xytext=(100, val_adm),
            arrowprops=dict(arrowstyle='<->', color='#D8C61C', lw=1.5, mutation_scale=10))
ax.text(120, (val_adm + val_adj) / 2, rf'CMRR: {cmrr_adj:.1f} dB@50Hz' + '\n' + rf'($\Delta$+{improvement:.1f} dB)',
        color='#B38000', fontsize=9, fontweight='bold', ha='left', va='center')

# --- 5. 坐标轴与网格设置 ---
ax.set_xlim(1e-2, 1e8)
ax.set_ylim(-100, 100)
ax.set_ylabel(r'Gain $A_{\mathrm{dm}}, A_{\mathrm{cm}}$ (dB)')
ax.set_xlabel(r'Frequency (Hz)')
ax.grid(True, which="major", ls="-", color='#C0C0C0', alpha=0.8, linewidth=0.8)
ax.grid(True, which="minor", ls=":", color='#E0E0E0', alpha=0.5, linewidth=0.5)

# 图例
ax.legend(loc='upper right', framealpha=0.9, edgecolor='#CCCCCC', fontsize=9)

plt.tight_layout()

# 保存高质量图像
plt.savefig('cmrr_two_arrows_plot.png', dpi=300, bbox_inches='tight')
plt.savefig('cmrr_two_arrows_plot.svg', format='svg', bbox_inches='tight')

plt.show()
```
